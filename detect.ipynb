```python
from ultralytics import YOLO
import os
from dotenv import load_dotenv
from roboflow import Roboflow
```

```python
load_dotenv()
API_KEY = os.environ['ROBOFLOW_TOKEN']
```

#### Load data

```python
rf = Roboflow(api_key=API_KEY)
dataset = rf.workspace("idcard-b708e").project("card_seg-4mofm").version(1).download("yolo26")
```

#### Load model

```python
model = YOLO('yolo26n-seg.pt')  
```

#### Train

```python
results = model.train(
    data=dataset.location,
    epochs=150,          
    imgsz=640,            
    batch=16,              
    patience=30,           
    optimizer='auto',      
    augment=True,
    degrees=5.0,           
    fliplr=0.0,            
    mosaic=0.0,            
    project='card_seg_runs',
    name='v1',
)
```

```python
model.export(format='onnx')
```
