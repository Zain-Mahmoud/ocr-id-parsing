```python
from ultralytics import YOLO
```

#### Load model

```python
model = YOLO('yolo26n.pt')  
```

#### Train

```python
results = model.train(
    data="./data/detection-data",
    epochs=150,          
    imgsz=640,            
    batch=16,              
    patience=30,           
    optimizer='auto',      
    augment=True,
    degrees=5.0,           
    fliplr=0.0,            
    mosaic=0.0,            
    project='card_seg_runs',
    name='v1',
)
```

```python
model.export(format='onnx')
```
