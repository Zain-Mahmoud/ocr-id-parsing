```python
from ultralytics import YOLO
```

```python
model = YOLO('yolo26n-seg.pt')  
```

```python
results = model.train(
    data='./data/card_seg.v1i.yolo26/data.yaml',
    epochs=150,          
    imgsz=640,            
    batch=16,              
    patience=30,           
    optimizer='auto',      
    augment=True,
    degrees=5.0,           
    fliplr=0.0,            
    mosaic=0.0,            
    project='card_seg_runs',
    name='v1',
)

model.export(format='onnx')
```
