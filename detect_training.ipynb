```python
from ultralytics import YOLO
import os
from dotenv import load_dotenv
from roboflow import Roboflow
```

```python
load_dotenv()
API_KEY = os.environ['ROBOFLOW_TOKEN']
```

#### Load data

```python
rf = Roboflow(api_key=API_KEY)
project = rf.workspace("zain-elsayed").project("ocr-extraction-0gzyi")
version = project.version(1)
dataset = version.download("yolo26")
                
```

#### Load model

```python
model = YOLO('yolo26s-obb.pt')  
```

#### Train

```python
results = model.train(
    data=dataset.location,
    epochs=150,          
    imgsz=960,            
    batch=16,              
    patience=25,           
    optimizer='auto',      
    augment=True,   
    fliplr=0.0,            
    mosaic=0.0,            
    project='card_obb_runs',
    name='v1',
)
```

```python
model.export(format='onnx')
```
